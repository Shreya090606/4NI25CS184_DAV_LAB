```python
import numpy as np
marks= np.array([78,85,94,96,90])
print("Marks:", marks)
print("Type:",type(marks))
```

```python
print("Number of dimensions:", marks.ndim)
print("Shape:", marks.shape)
print("Number of elements:", marks.size)
```

```python
marks= np.array([78,85,94,96,90])
print("Original Marks:", marks)
print("Add 5:", marks+5)
print("Multiply by 2:", marks*2)
print("Divide by 2:", marks/2)
print("Square:", marks**2)
```

```python
print("Sum:", marks.sum())
print("Mean:", marks.mean())
print("Maximun:", marks.max())
print("Minimun:", marks.min())
```

```python
print("First mark:", marks[0])
print("Second mark:", marks[1])
print("Last mark:", marks[-1])
```

```python
print("Index 1 to 3:", marks[1:4])
print("First three:", marks[:3])
print("From index 2:", marks[2:])
print("Every second element:", marks[ : :2])
```

```python
student_marks= np.array([[78,85,92],
                        [67,88,76],
                        [90,95,89]])
print(student_marks)
print("Shape:", student_marks.shape)
```

```python
print("Mean of all values:", student_marks.mean())
print("Column-wise mean:", student_marks.mean(axis=0))
print("Row-wise mean:", student_marks.mean(axis=1))
```

```python
import pandas as pd
data= {"Name": ["Anu", "Rahul","Kiran","Priya"],
      "Branch" : ["CSE", "AIML","CSE", "AIML"],
      "Marks" : [85,91,76,88]}
df=pd.DataFrame(data)
df
```

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns)
```

```python
df.head()
```

```python
df.tail()
```

```python
df.info()
```

```python
df.describe()
```

```python
print("Marks column:")
print(df["Marks"])
print("\n Branch and Marks:")
print(df[["Branch", "Marks"]])
```

```python
print("First row:")
print(df.iloc[0])
print("\nSecond row:")
print(df.iloc[1])
print("\nFirst three rows:")
print(df.iloc[0:3])
```

```python
print("First two rows and first two columns:")
print(df.iloc[0:2,0:2])
```

```python
print("First three rows:")
display(df.iloc[0:3])
print("Rows 1 to 3, Name and Marks:")
print(df.iloc[1:4,0:3])
```

```python
high_scores=df[df["Marks"]>90]
high_scores
```

```python
print("Students from CSE:")
display(df[df["Branch"] == "CSE"])
```

```python
print("Students scoring 85 or above:")
display(df[df["Marks"]>=85])
```
